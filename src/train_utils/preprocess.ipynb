```python
import os
import numpy as np
import random
```

```python

def collect_image_paths(root_dir, num_vids, seq_length, train_percent):
    image_paths = []
    
    # Collect all subdirectories (VID folders)
    vid_dirs = [os.path.join(root_dir, d) for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))]
    
    # Randomly select a subset of VID folders
    selected_vid_dirs = random.sample(vid_dirs, min(num_vids, len(vid_dirs)))
    
    # Randomly shuffle the selected VID folders
    random.shuffle(selected_vid_dirs)
    
    # Determine the number of VID folders for training and validation
    num_train_vids = int(train_percent * len(selected_vid_dirs))
    print("num_train_vids", num_train_vids)
    
    # Split selected VID folders into training and validation sets
    train_vids = selected_vid_dirs[:num_train_vids]
    print("train_vids", train_vids)
    val_vids = selected_vid_dirs[num_train_vids:]
    print("val_vids", val_vids)

    # Collect image paths for training and validation
    train_paths = []
    val_paths = []
    val = 0
    train = 0
    
    for vid_dir in train_vids:
        image_files = sorted([os.path.join(vid_dir, f) for f in os.listdir(vid_dir) if f.endswith('.png')])
        
        # Skip folders with fewer images than the sequence length
        if len(image_files) < seq_length:
            continue
        
        # Create non-overlapping sequences
        step = 1  # Adjust this step value to control overlap step
        for i in range(0, len(image_files) - seq_length + 1, step):
            sequences = image_files[i:i + seq_length]
            train_paths.append(sequences)
        train += 1
    
    for vid_dir in val_vids:
        image_files = sorted([os.path.join(vid_dir, f) for f in os.listdir(vid_dir) if f.endswith('.png')])
        
        # Skip folders with fewer images than the sequence length
        if len(image_files) < seq_length:
            continue
        
        # Create non-overlapping sequences
        step = 1  # Adjust this step value to control overlap step
        for i in range(0, len(image_files) - seq_length + 1, step):
            sequences = image_files[i:i + seq_length]
            val_paths.append(sequences)

        val += 1
    
    return train_paths, val_paths, val, train

```

```python
# Define the root directory of the dataset
root_dir = '/path/to/data'

# Set the number of VID directories (sub-directories) to include
num_vids = 1000 # For Catracts_1K dataset
num_vids = 80 # For Cholec80 dataset

# Set the sequence length (default is 3)
seq_length = 10 # 3 or 10

# Define the percentage for training (e.g., 70%)
train_percent = 0.7

# Collect image paths for training and validation
train_sequences, val_sequences, val, train = collect_image_paths(root_dir, num_vids, seq_length, train_percent)

# Save the training and validation sequences to Numpy files
np.save('/path/to/additional_files/train_seq_10_files.npy', train_sequences)
np.save('/path/to/additional_files/test_seq_10_files.npy', val_sequences)
```
